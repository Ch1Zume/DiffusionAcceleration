```python
def build_action_array(interval: int, length: int = 40):
    assert interval >= 1, "interval must be >= 1"

    arr = [1] * length

    # 每隔 interval 放一个 0
    for i in range(0, length, interval):
        arr[i] = 0

    # 最后两个强制为 0
    arr[-2:] = [0, 0]

    return arr

teacher_actions = build_action_array(interval=4, length=28)
print(teacher_actions)
```
